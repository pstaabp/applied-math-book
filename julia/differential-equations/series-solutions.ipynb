```julia
using CairoMakie, LaTeXStrings
CairoMakie.activate!()
Makie.inline!(true)
```

```julia
function S1(x::Real)
  a = -1
  s = 1-x
  for n=2:300
    a *= -1/(2n+1) /(n+1) 
    # @show a
    s += a*x^n
  end
  s
end
```

```julia
S1(1)
```

```julia
lines(0..100,S1)
```

```julia
function S2(x::Real)
  a = 1
  s = 1
  for n=0:5
    a *= -1/(1/2+n+1) /(2n+2) 
    # @show a
    s += a*x^n
  end
  s*sqrt(x)
end
```

```julia
S2(1)
```

```julia
fig, ax = lines(0..5,S2, label = L"y_2(x)")
lines!(ax,0..5,S1, label = L"y_1(x)")
axislegend(ax, position = (1,0.75))
fig
```
